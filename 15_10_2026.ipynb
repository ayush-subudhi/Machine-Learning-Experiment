```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import convolve
```

```python
img = cv2.imread("/content/images.jpeg",cv2.IMREAD_GRAYSCALE)
```

```python
plt.figure(figsize=(6,4))
plt.imshow(img, cmap="gray")
```

```python
sharpen=[[0,-1,0],
         [-1,5,-1],
         [0,-1,0]]
blur = np.ones((15,15),dtype=float)/10
```

```python
kernal=blur
#performing the convolution operation.

im=np.array(img,dtype=float)
img2=convolve(img,kernal)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2,cmap='gray')
```
